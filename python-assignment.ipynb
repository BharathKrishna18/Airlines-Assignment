```python
import pandas as pd

df = pd.read_excel('/kaggle/input/datasets/bharathkrishna18/airlines-dataset/UseCase - Airlines.xlsx')
print(df.shape)
```

Data Cleaning

```python
#Drop duplicates
df_updated = df.drop_duplicates()
print(df_updated.shape)
```

```python
print(df_updated.columns)
```

```python
#Handle empty values
df_updated['airline'] = df_updated['airline'].replace('',pd.NA)

mapping = df_updated.dropna(subset=['airline']).drop_duplicates('flight_id').set_index('flight_id')['airline']

df_updated['airline'] = df_updated['airline'].fillna(df['flight_id'].map(mapping))

print(df_updated.iloc[50:65])
```

```python
result = (df_updated['departure_time'] > df_updated['arrival_time'])
print(result)
```

```python
df_updated.loc[result,"arrival_time"] = (
    df_updated.loc[result,"arrival_time"] + pd.Timedelta(days=1)
)

print(df_updated.iloc[340:360])
```

```python
df_updated.to_csv('airlines_updated.csv', index=False)
```
